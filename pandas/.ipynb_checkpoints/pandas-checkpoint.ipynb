```python
#data frame 
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
```

```python
df= pd.DataFrame(data_list)
col = ["name","age","city","salary"]
df = pd.DataFrame(data_list, columns =col)
df
```

```python
#selection and indexing of columns 
```

```python
df["name"]
```

```python
df[["name","city"]]           #here we dont use index we use name instead 
```

#condition selection


```python
df[df["age"] > 30]
```

```python
df[(df["age"] > 30) & (df["city"] == "Paris") ]   #use proper bracket and & is used as and 
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
#removing missing data 
```

```python
df.dropna()
```

```python
df.dropna(thresh = 3)
```

```python
#filling the missing data 
```

```python
df.fillna(0)
```

```python
values = {'A':0,'B':100,"C":300,'D':400}
df.fillna(value=values)
```

```python
df.fillna(df.mean())
```
